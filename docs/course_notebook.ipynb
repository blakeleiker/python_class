### Lists

One of the most useful components of python are the **containers**. Containers are used to store objects.

**Lists** are the most commonly used type of container. They are mutable, ordered, and can contain objects of any type.

https://docs.python.org/3/tutorial/introduction.html#lists

```python
# Create a list where the first element is an int, the second is a string, and 
# the third is a float.
my_list = [0, 'cat', 3.14159]

print(my_list)
```

```python
# Lists may by changed after they are created.
# For example the append method adds a new element to the end of the list.
my_list.append('dog')

print(my_list)
```

### Tuples

Another type of container is a **tuple**.
Tuples are different from lists in that they are **immutable**, meaning they cannot be changed after they are created.

https://docs.python.org/3/tutorial/datastructures.html#tuples-and-sequences

```python
my_tuple = ('a', 'b', 'c')
print(my_tuple)
```

```python
my_tuple[0] = 'z'
```

Another commonly used container in python is the **dictionary**. A dictionary represents a set of key value pairs.

https://docs.python.org/3/tutorial/datastructures.html#dictionaries

```python
my_dict = {'a': 7, 'b': 12, 'c': [1, 2, 3]}
print(my_dict)
```

```python
# Retrieve the value associated with the key "a":
print(my_dict['a'])
```

### Numpy

[Numpy](https://numpy.org/doc/stable/user/absolute_beginners.html) is a commonly used scientific computing package for python. It is very useful for things like array or matrix manipulation, statistics, or linear algebra.

```python
import numpy as np

array = np.array([1.0, 5.0, 8.0, 1.234, 101.001])
print(array)
```

```python
# Compute the mean and standard deviation of this array
print( np.mean(array) )
print( np.std(array) )
```

```python
# define a 3x3 array
a = np.array([[1, 0, 0],
              [0, 0, 1],
              [0, 1, 0]])


print(a.shape)

# Compute the eigenvalues and eigenvectors of the array
np.linalg.eig(a)
```

### Pandas

[Pandas](https://pandas.pydata.org/docs/getting_started/index.html) is a data analysis package for python. Pandas provides a data type called a "DataFrame", which can be thought of as similar to an excel table. A DataFrame has an index, columns, and rows. 

```python
import pandas as pd

df = pd.DataFrame({
    'A': [1,2,3,4],
    'B': [5,6,7,8],
    'C': [9,10,11,12]
})

df
```

```python
# Add a new a new column by performing a transformation on another column
df['C Transformed'] = df['C'].transform(lambda x: x**x)

df
```
